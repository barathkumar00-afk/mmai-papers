```python
from pathlib import Path

base = Path(r"C:\Users\Windows\OneDrive\Desktop\mmai papers")

print("Searching for folders containing brain-tumor classes...\n")

target_names = {
    "glioma",
    "meningioma",
    "pituitary",
    "notumor",
    "no tumor"
}

for folder in base.rglob("*"):
    if folder.is_dir():
        try:
            children = {x.name.lower() for x in folder.iterdir() if x.is_dir()}
            
            matches = children.intersection(target_names)
            
            if len(matches) >= 2:
                print("📁 Possible dataset:", folder)
                print("   Classes found:", matches)
        except PermissionError:
            pass
```

```python
from pathlib import Path

desktop = Path(r"C:\Users\Windows\OneDrive\Desktop")

keywords = [
    "glioma",
    "meningioma",
    "pituitary",
    "notumor",
    "no_tumor",
    "brain_tumor",
    "brain tumor",
    "dataset"
]

print("Searching Desktop...\n")

found = set()

for path in desktop.rglob("*"):
    if path.is_dir():
        name = path.name.lower()

        if any(keyword in name for keyword in keywords):
            found.add(path)

for path in sorted(found):
    print("📁", path)

print(f"\nFound: {len(found)} matching folders")
```
